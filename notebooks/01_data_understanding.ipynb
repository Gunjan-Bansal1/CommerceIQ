```python
# Step 3.3: Column-level profiling

column_profiles = []

for dataset_name, df in datasets.items():

    for column in df.columns:

        series = df[column]

        # Exclude null, empty and whitespace-only values
        valid_mask = (
            series.notna()
            & series.str.strip().ne("").fillna(False)
        )

        valid_values = series[valid_mask]

        unique_count = valid_values.nunique()
        valid_count = len(valid_values)

        uniqueness_ratio = (
            unique_count / valid_count
            if valid_count > 0
            else None
        )

        # Classify column cardinality
        if unique_count == 0:
            cardinality = "No valid values"

        elif unique_count == 1:
            cardinality = "Constant"

        elif unique_count == valid_count:
            cardinality = "Unique"

        elif unique_count <= 20:
            cardinality = "Low"

        elif unique_count <= 100:
            cardinality = "Medium"

        else:
            cardinality = "High"

        column_profiles.append({
            "dataset": dataset_name,
            "column": column,
            "current_dtype": str(series.dtype),
            "total_records": len(series),
            "valid_records": valid_count,
            "unique_values": unique_count,
            "uniqueness_pct": (
                round(uniqueness_ratio * 100, 2)
                if uniqueness_ratio is not None
                else None
            ),
            "cardinality": cardinality,
            "sample_values": " | ".join(
                valid_values.drop_duplicates()
                .head(3)
                .str.slice(0, 80)
                .tolist()
            )
        })

column_profile_df = pd.DataFrame(column_profiles)

print("COMMERCEIQ - COLUMN LEVEL PROFILING")

# Display all column profiles
with pd.option_context(
    "display.max_rows", None,
    "display.max_colwidth", 100
):
    display(column_profile_df)

# Cardinality distribution
print("\nCARDINALITY DISTRIBUTION")

display(
    column_profile_df["cardinality"]
    .value_counts()
    .rename_axis("cardinality")
    .reset_index(name="column_count")
)

# Identify potential single-column unique keys
print("\nPOTENTIAL UNIQUE KEY CANDIDATES")

display(
    column_profile_df.loc[
        column_profile_df["cardinality"].eq("Unique")
        & column_profile_df["valid_records"].eq(
            column_profile_df["total_records"]
        ),
        [
            "dataset",
            "column",
            "total_records",
            "unique_values",
            "uniqueness_pct"
        ]
    ]
)

# Identify constant columns
print("\nCONSTANT COLUMNS")

display(
    column_profile_df.loc[
        column_profile_df["cardinality"].eq("Constant"),
        [
            "dataset",
            "column",
            "valid_records",
            "sample_values"
        ]
    ]
)

print(f"\nDatasets profiled: {len(datasets)}")
print(f"Total columns profiled: {len(column_profile_df)}")
```

```python
# Step 3.4: Missing value analysis

missing_records = []

for dataset_name, df in datasets.items():

    for column in df.columns:

        series = df[column]

        # Identify different forms of missing data
        null_count = int(series.isna().sum())

        empty_count = int(
            series.eq("").fillna(False).sum()
        )

        whitespace_count = int(
            (
                series.str.strip().eq("").fillna(False)
                & series.ne("").fillna(False)
            ).sum()
        )

        total_missing = (
            null_count
            + empty_count
            + whitespace_count
        )

        missing_pct = (
            total_missing / len(series) * 100
            if len(series) > 0
            else 0
        )

        # Classify missing value severity
        if missing_pct == 0:
            severity = "None"

        elif missing_pct <= 5:
            severity = "Low"

        elif missing_pct <= 20:
            severity = "Moderate"

        elif missing_pct <= 50:
            severity = "High"

        elif missing_pct < 100:
            severity = "Critical"

        else:
            severity = "Completely missing"

        missing_records.append({
            "dataset": dataset_name,
            "column": column,
            "total_records": len(series),
            "null_values": null_count,
            "empty_strings": empty_count,
            "whitespace_values": whitespace_count,
            "total_missing": total_missing,
            "missing_pct": round(missing_pct, 2),
            "completeness_pct": round(100 - missing_pct, 2),
            "severity": severity
        })

missing_profile_df = pd.DataFrame(missing_records)

# Display only columns containing missing values
missing_columns_df = (
    missing_profile_df[
        missing_profile_df["total_missing"] > 0
    ]
    .sort_values("missing_pct", ascending=False)
    .reset_index(drop=True)
)

print("COMMERCEIQ - MISSING VALUE ANALYSIS")

display(missing_columns_df)

# Dataset-level completeness summary
missing_summary_df = (
    missing_profile_df
    .groupby("dataset", as_index=False)
    .agg(
        total_cells=("total_records", "sum"),
        missing_cells=("total_missing", "sum"),
        total_columns=("column", "count"),
        affected_columns=(
            "total_missing",
            lambda x: x.gt(0).sum()
        )
    )
)

missing_summary_df["missing_pct"] = (
    missing_summary_df["missing_cells"]
    / missing_summary_df["total_cells"]
    * 100
).round(2)

missing_summary_df["completeness_pct"] = (
    100 - missing_summary_df["missing_pct"]
).round(2)

missing_summary_df = missing_summary_df.sort_values(
    "missing_pct",
    ascending=False
).reset_index(drop=True)

print("\nDATASET-LEVEL COMPLETENESS")

display(missing_summary_df)

# Missing value severity distribution
print("\nMISSING VALUE SEVERITY DISTRIBUTION")

display(
    missing_profile_df["severity"]
    .value_counts()
    .rename_axis("severity")
    .reset_index(name="column_count")
)

print(f"\nTotal columns analyzed: {len(missing_profile_df)}")
print(f"Columns with missing values: {len(missing_columns_df)}")
print(f"Complete columns: {len(missing_profile_df) - len(missing_columns_df)}")
print(f"Total missing cells: {missing_profile_df['total_missing'].sum():,}")
```

```python
# Step 3.5: Duplicate analysis

# 1. Full-row duplicate analysis

duplicate_records = []

for dataset_name, df in datasets.items():

    duplicate_count = int(df.duplicated().sum())

    affected_rows = int(
        df.duplicated(keep=False).sum()
    )

    duplicate_records.append({
        "dataset": dataset_name,
        "total_rows": len(df),
        "exact_duplicate_rows": duplicate_count,
        "rows_in_duplicate_groups": affected_rows,
        "distinct_rows": len(df) - duplicate_count,
        "duplicate_pct": round(
            duplicate_count / len(df) * 100, 4
        )
    })

duplicate_summary_df = (
    pd.DataFrame(duplicate_records)
    .sort_values("exact_duplicate_rows", ascending=False)
    .reset_index(drop=True)
)

print("COMMERCEIQ - EXACT DUPLICATE ANALYSIS")

display(duplicate_summary_df)


# 2. Business-key duplicate analysis

key_candidates = {
    "orders": [
        ["order_id"]
    ],

    "customers": [
        ["customer_id"],
        ["customer_unique_id"]
    ],

    "order_items": [
        ["order_id", "order_item_id"]
    ],

    "payments": [
        ["order_id", "payment_sequential"]
    ],

    "reviews": [
        ["review_id"],
        ["order_id"],
        ["review_id", "order_id"]
    ],

    "products": [
        ["product_id"]
    ],

    "sellers": [
        ["seller_id"]
    ],

    "geolocation": [
        ["geolocation_zip_code_prefix"]
    ],

    "category_translation": [
        ["product_category_name"]
    ],

    "marketing_leads": [
        ["mql_id"]
    ],

    "closed_deals": [
        ["mql_id"],
        ["seller_id"]
    ],

    "brazil_holidays": [
        ["date"],
        ["date", "countryCode"]
    ]
}

key_duplicate_records = []

for dataset_name, candidate_keys in key_candidates.items():

    df = datasets[dataset_name]

    for key_columns in candidate_keys:

        duplicate_count = int(
            df.duplicated(subset=key_columns).sum()
        )

        affected_rows = int(
            df.duplicated(
                subset=key_columns,
                keep=False
            ).sum()
        )

        distinct_keys = len(df) - duplicate_count

        key_duplicate_records.append({
            "dataset": dataset_name,
            "key_columns": " + ".join(key_columns),
            "total_rows": len(df),
            "distinct_keys": distinct_keys,
            "duplicate_key_rows": duplicate_count,
            "rows_in_duplicate_groups": affected_rows,
            "is_unique": duplicate_count == 0
        })

key_duplicate_df = pd.DataFrame(key_duplicate_records)

print("\nBUSINESS-KEY DUPLICATE ANALYSIS")

display(key_duplicate_df)


# 3. Consolidated findings

print("\nDATASETS CONTAINING EXACT DUPLICATES")

display(
    duplicate_summary_df[
        duplicate_summary_df["exact_duplicate_rows"] > 0
    ]
)

print("\nNON-UNIQUE BUSINESS KEY CANDIDATES")

display(
    key_duplicate_df[
        ~key_duplicate_df["is_unique"]
    ]
)

print(f"\nDatasets analyzed: {len(datasets)}")

print(
    "Total exact duplicate rows:",
    f"{duplicate_summary_df['exact_duplicate_rows'].sum():,}"
)

print(
    "Datasets containing exact duplicates:",
    duplicate_summary_df["exact_duplicate_rows"].gt(0).sum()
)

print(f"Business key combinations analyzed: {len(key_duplicate_df)}")
```

```python
# Step 3.6: Primary keys and dataset granularity

dataset_definitions = {
    "orders": {
        "grain": "One record per order",
        "primary_key": ["order_id"],
        "foreign_keys": ["customer_id"]
    },

    "customers": {
        "grain": "One record per order-specific customer ID",
        "primary_key": ["customer_id"],
        "foreign_keys": []
    },

    "order_items": {
        "grain": "One record per item within an order",
        "primary_key": ["order_id", "order_item_id"],
        "foreign_keys": ["order_id", "product_id", "seller_id"]
    },

    "payments": {
        "grain": "One record per payment sequence within an order",
        "primary_key": ["order_id", "payment_sequential"],
        "foreign_keys": ["order_id"]
    },

    "reviews": {
        "grain": "One record per review and order combination",
        "primary_key": ["review_id", "order_id"],
        "foreign_keys": ["order_id"]
    },

    "products": {
        "grain": "One record per product",
        "primary_key": ["product_id"],
        "foreign_keys": []
    },

    "sellers": {
        "grain": "One record per seller",
        "primary_key": ["seller_id"],
        "foreign_keys": []
    },

    "geolocation": {
        "grain": "One geographic observation per source row",
        "primary_key": [],
        "foreign_keys": []
    },

    "category_translation": {
        "grain": "One record per Portuguese product category",
        "primary_key": ["product_category_name"],
        "foreign_keys": []
    },

    "marketing_leads": {
        "grain": "One record per marketing-qualified lead",
        "primary_key": ["mql_id"],
        "foreign_keys": []
    },

    "closed_deals": {
        "grain": "One record per closed marketing lead",
        "primary_key": ["mql_id"],
        "foreign_keys": ["mql_id"]
    },

    "brazil_holidays": {
        "grain": "One holiday record per date in the current snapshot",
        "primary_key": ["date", "countryCode"],
        "foreign_keys": []
    }
}


# Validate proposed primary keys

granularity_records = []

for dataset_name, definition in dataset_definitions.items():

    df = datasets[dataset_name]
    primary_key = definition["primary_key"]

    if primary_key:

        key_name = " + ".join(primary_key)

        # Reuse previous duplicate analysis
        key_result = key_duplicate_df.loc[
            key_duplicate_df["dataset"].eq(dataset_name)
            & key_duplicate_df["key_columns"].eq(key_name)
        ].iloc[0]

        duplicate_keys = int(
            key_result["duplicate_key_rows"]
        )

        # Check whether any key component is missing
        missing_key_mask = pd.Series(
            False,
            index=df.index
        )

        for column in primary_key:

            missing_key_mask |= (
                df[column].isna()
                | df[column].str.strip().eq("").fillna(False)
            )

        missing_keys = int(missing_key_mask.sum())

        key_type = (
            "Single-column"
            if len(primary_key) == 1
            else "Composite"
        )

        key_status = (
            "VALID"
            if duplicate_keys == 0 and missing_keys == 0
            else "INVALID"
        )

    else:

        key_name = "No natural primary key"
        duplicate_keys = None
        missing_keys = None
        key_type = "Surrogate key required"
        key_status = "DESIGN REQUIRED"

    granularity_records.append({
        "dataset": dataset_name,
        "grain": definition["grain"],
        "proposed_primary_key": key_name,
        "key_type": key_type,
        "foreign_keys": ", ".join(
            definition["foreign_keys"]
        ) or "None",
        "total_rows": len(df),
        "duplicate_primary_keys": duplicate_keys,
        "missing_primary_keys": missing_keys,
        "key_status": key_status
    })


granularity_df = pd.DataFrame(granularity_records)

print("COMMERCEIQ - DATASET GRANULARITY AND KEY DESIGN")

display(granularity_df)

print("\nPRIMARY KEY VALIDATION SUMMARY")

display(
    granularity_df["key_status"]
    .value_counts()
    .rename_axis("key_status")
    .reset_index(name="dataset_count")
)

print(f"\nDatasets analyzed: {len(granularity_df)}")

print(
    "Valid proposed primary keys:",
    granularity_df["key_status"].eq("VALID").sum()
)

print(
    "Datasets requiring key design:",
    granularity_df["key_status"].eq("DESIGN REQUIRED").sum()
)

print(
    "Invalid proposed primary keys:",
    granularity_df["key_status"].eq("INVALID").sum()
)
```

```python
# Step 3.7: Dataset relationships and referential integrity

# Define relationships to investigate
relationships = [
    ("orders", "customer_id",
     "customers", "customer_id"),

    ("order_items", "order_id",
     "orders", "order_id"),

    ("order_items", "product_id",
     "products", "product_id"),

    ("order_items", "seller_id",
     "sellers", "seller_id"),

    ("payments", "order_id",
     "orders", "order_id"),

    ("reviews", "order_id",
     "orders", "order_id"),

    ("products", "product_category_name",
     "category_translation", "product_category_name"),

    ("customers", "customer_zip_code_prefix",
     "geolocation", "geolocation_zip_code_prefix"),

    ("sellers", "seller_zip_code_prefix",
     "geolocation", "geolocation_zip_code_prefix"),

    ("closed_deals", "mql_id",
     "marketing_leads", "mql_id"),

    ("closed_deals", "seller_id",
     "sellers", "seller_id")
]


relationship_records = []

unmatched_key_records = []

for child_name, child_key, parent_name, parent_key in relationships:

    child = datasets[child_name][child_key]
    parent = datasets[parent_name][parent_key]

    # Exclude missing keys
    child_valid = child[
        child.notna()
        & child.str.strip().ne("").fillna(False)
    ]

    parent_valid = parent[
        parent.notna()
        & parent.str.strip().ne("").fillna(False)
    ]

    # Distinct keys
    child_keys = set(child_valid.unique())
    parent_keys = set(parent_valid.unique())

    # Determine observed join cardinality
    child_unique = child_valid.is_unique
    parent_unique = parent_valid.is_unique

    if child_unique and parent_unique:
        cardinality = "One-to-one"

    elif not child_unique and parent_unique:
        cardinality = "Many-to-one"

    elif child_unique and not parent_unique:
        cardinality = "One-to-many"

    else:
        cardinality = "Many-to-many"

    # Match child records against parent records
    matched_mask = child_valid.isin(parent_keys)

    matched_rows = int(matched_mask.sum())

    unmatched_rows = int((~matched_mask).sum())

    unmatched_keys = child_keys - parent_keys

    missing_child_keys = len(child) - len(child_valid)

    # Identify parent keys without corresponding child records
    unused_parent_keys = parent_keys - child_keys

    match_pct = (
        matched_rows / len(child_valid) * 100
        if len(child_valid) > 0
        else None
    )

    relationship_records.append({
        "child_dataset": child_name,
        "child_key": child_key,
        "parent_dataset": parent_name,
        "parent_key": parent_key,
        "cardinality": cardinality,
        "child_rows": len(child),
        "missing_child_keys": missing_child_keys,
        "matched_rows": matched_rows,
        "unmatched_rows": unmatched_rows,
        "unmatched_distinct_keys": len(unmatched_keys),
        "unused_parent_keys": len(unused_parent_keys),
        "match_pct": round(match_pct, 4)
        if match_pct is not None else None,
        "status": (
            "COMPLETE MATCH"
            if unmatched_rows == 0 and missing_child_keys == 0
            else "REVIEW REQUIRED"
        )
    })

    # Record unmatched keys for subsequent cleaning
    for key in sorted(unmatched_keys):

        unmatched_key_records.append({
            "child_dataset": child_name,
            "child_key": child_key,
            "parent_dataset": parent_name,
            "unmatched_key": key
        })


relationship_df = pd.DataFrame(relationship_records)

unmatched_keys_df = pd.DataFrame(
    unmatched_key_records,
    columns=[
        "child_dataset",
        "child_key",
        "parent_dataset",
        "unmatched_key"
    ]
)


print("COMMERCEIQ - DATASET RELATIONSHIP ANALYSIS")

display(relationship_df)


print("\nRELATIONSHIPS REQUIRING REVIEW")

display(
    relationship_df.loc[
        relationship_df["status"].eq("REVIEW REQUIRED"),
        [
            "child_dataset",
            "parent_dataset",
            "missing_child_keys",
            "unmatched_rows",
            "unmatched_distinct_keys",
            "match_pct"
        ]
    ]
)


print("\nJOIN CARDINALITY SUMMARY")

display(
    relationship_df["cardinality"]
    .value_counts()
    .rename_axis("cardinality")
    .reset_index(name="relationship_count")
)


print("\nCATEGORY TRANSLATIONS NOT AVAILABLE")

display(
    unmatched_keys_df.loc[
        unmatched_keys_df["child_dataset"].eq("products")
        & unmatched_keys_df["parent_dataset"].eq(
            "category_translation"
        )
    ]
)


print("\nRELATIONSHIP VALIDATION SUMMARY")

print(f"Relationships analyzed: {len(relationship_df)}")

print(
    "Complete matches:",
    relationship_df["status"].eq("COMPLETE MATCH").sum()
)

print(
    "Relationships requiring review:",
    relationship_df["status"].eq("REVIEW REQUIRED").sum()
)
```

```python
# Step 3.8: Numerical profiling and outlier detection

import numpy as np

# Define numerical columns based on business meaning
numeric_columns = {
    "order_items": [
        "order_item_id",
        "price",
        "freight_value"
    ],

    "payments": [
        "payment_sequential",
        "payment_installments",
        "payment_value"
    ],

    "reviews": [
        "review_score"
    ],

    "products": [
        "product_name_lenght",
        "product_description_lenght",
        "product_photos_qty",
        "product_weight_g",
        "product_length_cm",
        "product_height_cm",
        "product_width_cm"
    ],

    "geolocation": [
        "geolocation_lat",
        "geolocation_lng"
    ],

    "closed_deals": [
        "declared_product_catalog_size",
        "declared_monthly_revenue"
    ]
}


# Continuous measures suitable for IQR screening
iqr_columns = {
    "price",
    "freight_value",
    "payment_value",
    "product_name_lenght",
    "product_description_lenght",
    "product_photos_qty",
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm",
    "declared_product_catalog_size",
    "declared_monthly_revenue"
}


# Proposed business and technical boundaries
numeric_rules = {
    ("order_items", "order_item_id"): (1, None),
    ("order_items", "price"): (0, None),
    ("order_items", "freight_value"): (0, None),

    ("payments", "payment_sequential"): (1, None),
    ("payments", "payment_installments"): (0, None),
    ("payments", "payment_value"): (0, None),

    ("reviews", "review_score"): (1, 5),

    ("products", "product_name_lenght"): (0, None),
    ("products", "product_description_lenght"): (0, None),
    ("products", "product_photos_qty"): (0, None),
    ("products", "product_weight_g"): (0, None),
    ("products", "product_length_cm"): (0, None),
    ("products", "product_height_cm"): (0, None),
    ("products", "product_width_cm"): (0, None),

    ("geolocation", "geolocation_lat"): (-90, 90),
    ("geolocation", "geolocation_lng"): (-180, 180),

    ("closed_deals", "declared_product_catalog_size"): (0, None),
    ("closed_deals", "declared_monthly_revenue"): (0, None)
}


integer_columns = {
    "order_item_id",
    "payment_sequential",
    "payment_installments",
    "review_score",
    "product_name_lenght",
    "product_description_lenght",
    "product_photos_qty",
    "declared_product_catalog_size"
}


numeric_records = []

for dataset_name, columns in numeric_columns.items():

    df = datasets[dataset_name]

    for column in columns:

        raw = df[column]

        # Identify missing source values
        missing_mask = (
            raw.isna()
            | raw.str.strip().eq("").fillna(False)
        )

        # Convert temporarily without modifying source data
        numeric = pd.to_numeric(
            raw,
            errors="coerce"
        )

        finite_mask = (
            numeric.notna()
            & np.isfinite(numeric)
        )

        invalid_numeric_mask = (
            ~missing_mask
            & ~finite_mask
        )

        valid = numeric[finite_mask].astype(float)

        valid_count = len(valid)

        # Descriptive statistics
        q1 = valid.quantile(0.25)
        median = valid.median()
        q3 = valid.quantile(0.75)

        iqr = q3 - q1

        lower_fence = np.nan
        upper_fence = np.nan
        outlier_count = np.nan
        outlier_pct = np.nan

        # Apply IQR only to selected measures
        if column not in iqr_columns:

            iqr_status = "NOT APPLICABLE"

        elif valid_count == 0:

            iqr_status = "NO VALID VALUES"

        elif iqr == 0:

            iqr_status = "ZERO IQR - REVIEW"

        else:

            lower_fence = q1 - 1.5 * iqr
            upper_fence = q3 + 1.5 * iqr

            outlier_mask = (
                (valid < lower_fence)
                | (valid > upper_fence)
            )

            outlier_count = int(outlier_mask.sum())

            outlier_pct = round(
                outlier_count / valid_count * 100,
                2
            )

            iqr_status = "CALCULATED"

        # Check business rule violations
        minimum_allowed, maximum_allowed = numeric_rules[
            (dataset_name, column)
        ]

        invalid_range_mask = pd.Series(
            False,
            index=valid.index
        )

        if minimum_allowed is not None:

            invalid_range_mask |= (
                valid < minimum_allowed
            )

        if maximum_allowed is not None:

            invalid_range_mask |= (
                valid > maximum_allowed
            )

        range_violations = int(
            invalid_range_mask.sum()
        )

        # Validate fields expected to contain integers
        non_integer_count = 0

        if column in integer_columns:

            non_integer_count = int(
                (valid % 1 != 0).sum()
            )

        numeric_records.append({
            "dataset": dataset_name,
            "column": column,
            "total_records": len(raw),
            "valid_numeric_records": valid_count,
            "missing_values": int(missing_mask.sum()),
            "invalid_numeric_values": int(
                invalid_numeric_mask.sum()
            ),
            "zero_values": int(valid.eq(0).sum()),
            "negative_values": int(valid.lt(0).sum()),
            "non_integer_values": non_integer_count,
            "minimum": valid.min(),
            "maximum": valid.max(),
            "mean": valid.mean(),
            "median": median,
            "std_dev": valid.std(),
            "q1": q1,
            "q3": q3,
            "iqr": iqr,
            "lower_fence": lower_fence,
            "upper_fence": upper_fence,
            "potential_outliers": outlier_count,
            "outlier_pct": outlier_pct,
            "iqr_status": iqr_status,
            "range_violations": range_violations
        })


numeric_profile_df = pd.DataFrame(
    numeric_records
).round(4)


# 1. Descriptive statistics

print("COMMERCEIQ - NUMERICAL PROFILING")

display(
    numeric_profile_df[
        [
            "dataset",
            "column",
            "valid_numeric_records",
            "missing_values",
            "minimum",
            "maximum",
            "mean",
            "median",
            "std_dev"
        ]
    ]
)


# 2. Numerical data quality

print("\nNUMERICAL DATA QUALITY CHECKS")

display(
    numeric_profile_df[
        [
            "dataset",
            "column",
            "invalid_numeric_values",
            "zero_values",
            "negative_values",
            "non_integer_values",
            "range_violations"
        ]
    ]
)


# 3. Potential statistical outliers

print("\nIQR OUTLIER ANALYSIS")

display(
    numeric_profile_df.loc[
        numeric_profile_df["column"].isin(iqr_columns),
        [
            "dataset",
            "column",
            "q1",
            "median",
            "q3",
            "iqr",
            "lower_fence",
            "upper_fence",
            "potential_outliers",
            "outlier_pct",
            "iqr_status"
        ]
    ]
)


# 4. Invalid numeric records

numeric_issues_df = numeric_profile_df.loc[
    numeric_profile_df["invalid_numeric_values"].gt(0)
    | numeric_profile_df["range_violations"].gt(0)
    | numeric_profile_df["non_integer_values"].gt(0)
].copy()

print("\nNUMERICAL FIELDS REQUIRING REVIEW")

display(
    numeric_issues_df[
        [
            "dataset",
            "column",
            "invalid_numeric_values",
            "range_violations",
            "non_integer_values"
        ]
    ]
)


print("\nNUMERICAL PROFILING SUMMARY")

print(f"Numerical columns analyzed: {len(numeric_profile_df)}")

print(
    "Invalid numeric values:",
    numeric_profile_df["invalid_numeric_values"].sum()
)

print(
    "Range violations:",
    numeric_profile_df["range_violations"].sum()
)

print(
    "Fields requiring review:",
    len(numeric_issues_df)
)

print("\nNumerical profiling completed.")
```

```python
# Step 3.9: Categorical and text profiling

import unicodedata

# 1. Define categorical columns relevant to business analysis

categorical_columns = {
    "orders": [
        "order_status"
    ],

    "customers": [
        "customer_city",
        "customer_state"
    ],

    "payments": [
        "payment_type"
    ],

    "reviews": [
        "review_score"
    ],

    "products": [
        "product_category_name"
    ],

    "sellers": [
        "seller_city",
        "seller_state"
    ],

    "geolocation": [
        "geolocation_city",
        "geolocation_state"
    ],

    "category_translation": [
        "product_category_name",
        "product_category_name_english"
    ],

    "marketing_leads": [
        "origin"
    ],

    "closed_deals": [
        "business_segment",
        "lead_type",
        "lead_behaviour_profile",
        "has_company",
        "has_gtin",
        "average_stock",
        "business_type"
    ],

    "brazil_holidays": [
        "localName",
        "name",
        "countryCode",
        "fixed",
        "global",
        "counties",
        "types"
    ]
}


# 2. Categorical frequency analysis

categorical_records = []
frequency_records = []

for dataset_name, columns in categorical_columns.items():

    for column in columns:

        series = datasets[dataset_name][column]

        missing_mask = (
            series.isna()
            | series.str.strip().eq("").fillna(False)
        )

        valid = series[~missing_mask]

        frequencies = valid.value_counts()

        unique_count = len(frequencies)

        most_frequent = (
            frequencies.index[0]
            if unique_count > 0
            else None
        )

        top_count = (
            int(frequencies.iloc[0])
            if unique_count > 0
            else 0
        )

        categorical_records.append({
            "dataset": dataset_name,
            "column": column,
            "total_records": len(series),
            "valid_records": len(valid),
            "missing_records": int(missing_mask.sum()),
            "unique_categories": unique_count,
            "most_frequent_value": most_frequent,
            "most_frequent_count": top_count,
            "most_frequent_pct": round(
                top_count / len(valid) * 100, 2
            ) if len(valid) else None,
            "single_occurrence_categories": int(
                frequencies.eq(1).sum()
            )
        })

        # Store complete category frequencies
        for category, count in frequencies.items():

            frequency_records.append({
                "dataset": dataset_name,
                "column": column,
                "category": category,
                "count": int(count),
                "percentage": round(
                    count / len(valid) * 100, 4
                )
            })


categorical_profile_df = pd.DataFrame(
    categorical_records
)

category_frequency_df = pd.DataFrame(
    frequency_records
)

print("COMMERCEIQ - CATEGORICAL PROFILING")

display(categorical_profile_df)


# 3. Display important business distributions

important_distributions = [
    ("orders", "order_status"),
    ("payments", "payment_type"),
    ("reviews", "review_score"),
    ("marketing_leads", "origin"),
    ("closed_deals", "business_type"),
    ("brazil_holidays", "types")
]

for dataset_name, column in important_distributions:

    print(f"\nDISTRIBUTION: {dataset_name}.{column}")

    display(
        category_frequency_df.loc[
            category_frequency_df["dataset"].eq(dataset_name)
            & category_frequency_df["column"].eq(column),
            ["category", "count", "percentage"]
        ].reset_index(drop=True)
    )


# 4. Detect inconsistent category spellings

def normalize_text(value):
    """Normalize text temporarily for profiling."""

    value = str(value).strip().casefold()

    value = unicodedata.normalize("NFKD", value)

    value = "".join(
        character
        for character in value
        if not unicodedata.combining(character)
    )

    return " ".join(value.split())


text_normalization_columns = [
    ("customers", "customer_city"),
    ("sellers", "seller_city"),
    ("geolocation", "geolocation_city"),
    ("products", "product_category_name"),
    ("marketing_leads", "origin"),
    ("closed_deals", "business_segment")
]

normalization_records = []
label_variant_records = []

for dataset_name, column in text_normalization_columns:

    series = datasets[dataset_name][column]

    valid = series[
        series.notna()
        & series.str.strip().ne("").fillna(False)
    ]

    frequencies = valid.value_counts()

    unique_labels = pd.DataFrame({
        "original_label": frequencies.index,
        "record_count": frequencies.values
    })

    unique_labels["normalized_label"] = (
        unique_labels["original_label"]
        .map(normalize_text)
    )

    # Identify labels affected by normalization
    changed_mask = unique_labels["original_label"].ne(
        unique_labels["normalized_label"]
    )

    # Detect different original labels sharing a normalized form
    variant_mask = unique_labels[
        "normalized_label"
    ].duplicated(keep=False)

    variants = unique_labels[variant_mask].copy()

    normalization_records.append({
        "dataset": dataset_name,
        "column": column,
        "original_unique_labels": len(unique_labels),
        "normalized_unique_labels":
            unique_labels["normalized_label"].nunique(),
        "labels_affected": int(changed_mask.sum()),
        "records_affected": int(
            unique_labels.loc[
                changed_mask, "record_count"
            ].sum()
        ),
        "variant_groups":
            variants["normalized_label"].nunique()
    })

    if not variants.empty:

        variants.insert(0, "column", column)
        variants.insert(0, "dataset", dataset_name)

        label_variant_records.extend(
            variants.to_dict("records")
        )


normalization_profile_df = pd.DataFrame(
    normalization_records
)

label_variants_df = pd.DataFrame(
    label_variant_records,
    columns=[
        "dataset",
        "column",
        "original_label",
        "record_count",
        "normalized_label"
    ]
)

print("\nTEXT NORMALIZATION ANALYSIS")

display(normalization_profile_df)

print("\nEXAMPLES OF INCONSISTENT LABELS")

display(
    label_variants_df.sort_values(
        ["dataset", "normalized_label"]
    ).head(20)
)


# 5. Boolean field validation

boolean_columns = [
    ("closed_deals", "has_company"),
    ("closed_deals", "has_gtin"),
    ("brazil_holidays", "fixed"),
    ("brazil_holidays", "global")
]

boolean_records = []

for dataset_name, column in boolean_columns:

    series = datasets[dataset_name][column]

    missing_mask = (
        series.isna()
        | series.str.strip().eq("").fillna(False)
    )

    normalized = series.str.strip().str.lower()

    true_count = int(normalized.eq("true").sum())
    false_count = int(normalized.eq("false").sum())

    invalid_count = int(
        (
            ~missing_mask
            & ~normalized.isin(["true", "false"])
        ).sum()
    )

    boolean_records.append({
        "dataset": dataset_name,
        "column": column,
        "true_count": true_count,
        "false_count": false_count,
        "missing_count": int(missing_mask.sum()),
        "invalid_count": invalid_count
    })


boolean_profile_df = pd.DataFrame(boolean_records)

print("\nBOOLEAN FIELD VALIDATION")

display(boolean_profile_df)


# 6. Review text profiling

text_records = []

for column in [
    "review_comment_title",
    "review_comment_message"
]:

    series = datasets["reviews"][column]

    valid = series[
        series.notna()
        & series.str.strip().ne("").fillna(False)
    ].str.strip()

    character_lengths = valid.str.len()

    word_counts = valid.str.split().str.len()

    text_records.append({
        "column": column,
        "total_reviews": len(series),
        "reviews_with_text": len(valid),
        "reviews_without_text": len(series) - len(valid),
        "text_availability_pct": round(
            len(valid) / len(series) * 100, 2
        ),
        "unique_text_values": valid.nunique(),
        "min_characters": character_lengths.min(),
        "max_characters": character_lengths.max(),
        "mean_characters": round(
            character_lengths.mean(), 2
        ),
        "median_characters": character_lengths.median(),
        "mean_words": round(word_counts.mean(), 2),
        "median_words": word_counts.median(),
        "max_words": word_counts.max()
    })


text_profile_df = pd.DataFrame(text_records)

print("\nREVIEW TEXT PROFILING")

display(text_profile_df)


# 7. Final summary

print("\nCATEGORICAL AND TEXT PROFILING SUMMARY")

print(
    "Categorical columns analyzed:",
    len(categorical_profile_df)
)

print(
    "Category frequency records:",
    len(category_frequency_df)
)

print(
    "Text normalization fields:",
    len(normalization_profile_df)
)

print(
    "Boolean fields analyzed:",
    len(boolean_profile_df)
)

print(
    "Invalid Boolean values:",
    boolean_profile_df["invalid_count"].sum()
)

print(
    "Review text fields analyzed:",
    len(text_profile_df)
)

print("\nCategorical and text profiling completed.")
```

```python
# Step 3.10: Temporal profiling and chronological validation

# 1. Define date and timestamp columns

date_columns = {
    "orders": [
        "order_purchase_timestamp",
        "order_approved_at",
        "order_delivered_carrier_date",
        "order_delivered_customer_date",
        "order_estimated_delivery_date"
    ],

    "order_items": [
        "shipping_limit_date"
    ],

    "reviews": [
        "review_creation_date",
        "review_answer_timestamp"
    ],

    "marketing_leads": [
        "first_contact_date"
    ],

    "closed_deals": [
        "won_date"
    ],

    "brazil_holidays": [
        "date"
    ]
}


# 2. Parse dates temporarily and profile temporal coverage

temporal_records = []
parsed_dates = {}

for dataset_name, columns in date_columns.items():

    parsed_dates[dataset_name] = {}

    for column in columns:

        raw = datasets[dataset_name][column]

        missing_mask = (
            raw.isna()
            | raw.str.strip().eq("").fillna(False)
        )

        converted = pd.to_datetime(
            raw,
            errors="coerce",
            format="mixed"
        )

        parsed_dates[dataset_name][column] = converted

        invalid_mask = (
            ~missing_mask
            & converted.isna()
        )

        valid = converted.dropna()

        temporal_records.append({
            "dataset": dataset_name,
            "column": column,
            "total_records": len(raw),
            "valid_dates": len(valid),
            "missing_dates": int(missing_mask.sum()),
            "invalid_dates": int(invalid_mask.sum()),
            "earliest_date": valid.min(),
            "latest_date": valid.max(),
            "distinct_dates": valid.dt.normalize().nunique(),
            "coverage_days": (
                (valid.max().normalize()
                 - valid.min().normalize()).days + 1
                if len(valid) else None
            )
        })


temporal_profile_df = pd.DataFrame(temporal_records)

print("COMMERCEIQ - TEMPORAL PROFILING")

display(temporal_profile_df)


# 3. Chronological consistency checks

chronology_records = []


def check_chronology(rule, earlier, later):
    """Check whether a later event occurs before an earlier event."""

    comparable = earlier.notna() & later.notna()

    violations = comparable & later.lt(earlier)

    comparable_count = int(comparable.sum())
    violation_count = int(violations.sum())

    return {
        "validation_rule": rule,
        "comparable_records": comparable_count,
        "records_not_comparable": int((~comparable).sum()),
        "violations": violation_count,
        "violation_pct": round(
            violation_count / comparable_count * 100, 4
        ) if comparable_count else None
    }


order_dates = parsed_dates["orders"]

purchase = order_dates["order_purchase_timestamp"]
approval = order_dates["order_approved_at"]
carrier = order_dates["order_delivered_carrier_date"]
delivery = order_dates["order_delivered_customer_date"]
estimated = order_dates["order_estimated_delivery_date"]


order_rules = [
    ("Approval before purchase", purchase, approval),
    ("Carrier handover before purchase", purchase, carrier),
    ("Carrier handover before approval", approval, carrier),
    ("Customer delivery before purchase", purchase, delivery),
    ("Customer delivery before approval", approval, delivery),
    ("Customer delivery before carrier handover", carrier, delivery),
    ("Estimated delivery date before purchase date",
     purchase.dt.normalize(), estimated.dt.normalize())
]

for rule, earlier, later in order_rules:

    chronology_records.append(
        check_chronology(rule, earlier, later)
    )


# Review chronology

chronology_records.append(
    check_chronology(
        "Review answer before review creation",
        parsed_dates["reviews"]["review_creation_date"],
        parsed_dates["reviews"]["review_answer_timestamp"]
    )
)


# Marketing chronology

lead_contact_lookup = pd.Series(
    parsed_dates["marketing_leads"]["first_contact_date"].values,
    index=datasets["marketing_leads"]["mql_id"]
)

closed_deal_contact = (
    datasets["closed_deals"]["mql_id"]
    .map(lead_contact_lookup)
)

chronology_records.append(
    check_chronology(
        "Deal won before first contact",
        closed_deal_contact,
        parsed_dates["closed_deals"]["won_date"]
    )
)


# Shipping deadline chronology

purchase_lookup = pd.Series(
    purchase.values,
    index=datasets["orders"]["order_id"]
)

item_purchase = (
    datasets["order_items"]["order_id"]
    .map(purchase_lookup)
)

chronology_records.append(
    check_chronology(
        "Shipping deadline before order purchase",
        item_purchase,
        parsed_dates["order_items"]["shipping_limit_date"]
    )
)


chronology_profile_df = pd.DataFrame(
    chronology_records
)

print("\nCHRONOLOGICAL CONSISTENCY CHECKS")

display(chronology_profile_df)


# 4. Order lifecycle duration profiling

duration_metrics = {
    "Purchase to approval (hours)":
        (approval - purchase).dt.total_seconds() / 3600,

    "Approval to carrier handover (days)":
        (carrier - approval).dt.total_seconds() / 86400,

    "Carrier handover to delivery (days)":
        (delivery - carrier).dt.total_seconds() / 86400,

    "Purchase to customer delivery (days)":
        (delivery - purchase).dt.total_seconds() / 86400,

    "Purchase date to estimated delivery date (days)":
        (estimated.dt.normalize()
         - purchase.dt.normalize()).dt.days,

    "Delivery date deviation from estimate (days)":
        (delivery.dt.normalize()
         - estimated.dt.normalize()).dt.days,

    "First contact to deal won (days)":
        (
            parsed_dates["closed_deals"]["won_date"]
            - closed_deal_contact
        ).dt.total_seconds() / 86400
}


duration_records = []

for metric, series in duration_metrics.items():

    valid = series.dropna()

    duration_records.append({
        "metric": metric,
        "valid_records": len(valid),
        "missing_records": int(series.isna().sum()),
        "minimum": valid.min(),
        "maximum": valid.max(),
        "mean": valid.mean(),
        "median": valid.median(),
        "negative_values": int(valid.lt(0).sum()),
        "zero_values": int(valid.eq(0).sum())
    })


duration_profile_df = pd.DataFrame(
    duration_records
).round(2)

print("\nLIFECYCLE DURATION PROFILING")

display(duration_profile_df)


# 5. Missing lifecycle timestamps by order status

status_date_df = pd.DataFrame({
    "order_status": datasets["orders"]["order_status"],
    "missing_approval": approval.isna(),
    "missing_carrier_date": carrier.isna(),
    "missing_delivery_date": delivery.isna()
})

status_timestamp_profile_df = (
    status_date_df
    .groupby("order_status", as_index=False)
    .agg(
        total_orders=("order_status", "size"),
        missing_approval=("missing_approval", "sum"),
        missing_carrier_date=("missing_carrier_date", "sum"),
        missing_delivery_date=("missing_delivery_date", "sum")
    )
    .sort_values("total_orders", ascending=False)
)

print("\nMISSING TIMESTAMPS BY ORDER STATUS")

display(status_timestamp_profile_df)


# 6. Monthly order coverage

monthly_order_profile_df = (
    purchase.dropna()
    .dt.to_period("M")
    .value_counts()
    .sort_index()
    .rename_axis("order_month")
    .reset_index(name="order_count")
)

monthly_order_profile_df["order_month"] = (
    monthly_order_profile_df["order_month"].astype(str)
)

print("\nMONTHLY ORDER COVERAGE")

display(monthly_order_profile_df)


# 7. Final summary

print("\nTEMPORAL PROFILING SUMMARY")

print(
    "Date columns analyzed:",
    len(temporal_profile_df)
)

print(
    "Invalid date values:",
    temporal_profile_df["invalid_dates"].sum()
)

print(
    "Chronological rules evaluated:",
    len(chronology_profile_df)
)

print(
    "Rules with violations:",
    chronology_profile_df["violations"].gt(0).sum()
)

print(
    "Duration metrics analyzed:",
    len(duration_profile_df)
)

print(
    "Observed order months:",
    len(monthly_order_profile_df)
)

print("\nTemporal profiling completed.")
```

```python
# Step 3.12: Generate consolidated Excel profiling report

from openpyxl.styles import Font, PatternFill, Alignment
from openpyxl.chart import BarChart, Reference
from openpyxl.formatting.rule import ColorScaleRule
from openpyxl.utils import get_column_letter


# 1. Executive summary

executive_summary_df = pd.DataFrame([
    ("Project", "CommerceIQ"),
    ("Phase", "3 - Data Understanding"),
    ("Source datasets", len(datasets)),
    ("Total source records", sum(len(df) for df in datasets.values())),
    ("Source columns", len(column_profile_df)),
    ("Columns with missing values", len(missing_columns_df)),
    ("Complete columns", int(
        missing_profile_df["total_missing"].eq(0).sum()
    )),
    ("Total missing cells", int(
        missing_profile_df["total_missing"].sum()
    )),
    ("Exact duplicate rows", int(
        duplicate_summary_df["exact_duplicate_rows"].sum()
    )),
    ("Valid proposed primary keys", int(
        granularity_df["key_status"].eq("VALID").sum()
    )),
    ("Relationships analyzed", len(relationship_df)),
    ("Complete relationship matches", int(
        relationship_df["status"].eq("COMPLETE MATCH").sum()
    )),
    ("Relationships requiring review", int(
        relationship_df["status"].eq("REVIEW REQUIRED").sum()
    )),
    ("Numerical columns profiled", len(numeric_profile_df)),
    ("Invalid numerical values", int(
        numeric_profile_df["invalid_numeric_values"].sum()
    )),
    ("Categorical columns profiled", len(categorical_profile_df)),
    ("Date columns profiled", len(temporal_profile_df)),
    ("Invalid date values", int(
        temporal_profile_df["invalid_dates"].sum()
    )),
    ("Chronological rules with violations", int(
        chronology_profile_df["violations"].gt(0).sum()
    )),
    ("Quality observations", len(data_quality_issues_df)),
    ("High-priority observations", int(
        data_quality_issues_df["priority"].eq("High").sum()
    )),
    ("Medium-priority observations", int(
        data_quality_issues_df["priority"].eq("Medium").sum()
    )),
    ("Low-priority observations", int(
        data_quality_issues_df["priority"].eq("Low").sum()
    )),
    ("Informational observations", int(
        data_quality_issues_df["priority"].eq("Informational").sum()
    )),
    ("Proposed core reporting period", "Jan 2017 - Aug 2018"),
    ("Raw data modified", "No"),
    ("Next phase", "Phase 4 - Data Cleaning")
], columns=["Metric", "Value"])


# 2. Select existing profiling outputs
# Exclude raw sample values from the Excel report.

report_tables = {
    "Executive Summary": executive_summary_df,
    "Dataset Overview": structure_df,
    "Data Dictionary": data_dictionary_df,

    "Column Profiling": column_profile_df.drop(
        columns=["sample_values"], errors="ignore"
    ),

    "Missing Values": missing_profile_df,
    "Completeness": missing_summary_df,
    "Exact Duplicates": duplicate_summary_df,
    "Key Analysis": key_duplicate_df,
    "Granularity": granularity_df,
    "Relationships": relationship_df,
    "Numerical Profiling": numeric_profile_df,
    "Categorical Profiling": categorical_profile_df,
    "Category Frequencies": category_frequency_df,
    "Text Normalization": normalization_profile_df,
    "Boolean Validation": boolean_profile_df,
    "Review Text": text_profile_df,
    "Temporal Profiling": temporal_profile_df,
    "Chronology Checks": chronology_profile_df,
    "Lifecycle Duration": duration_profile_df,
    "Status Timestamps": status_timestamp_profile_df,
    "Monthly Coverage": monthly_order_profile_df,
    "Quality Issues": data_quality_issues_df
}


# 3. Define workbook location

REPORT_DIR = PROJECT_ROOT / "reports"
REPORT_DIR.mkdir(parents=True, exist_ok=True)

PROFILING_REPORT_PATH = (
    REPORT_DIR / "data_profiling_report.xlsx"
)


# 4. Define formatting styles

header_fill = PatternFill(
    fill_type="solid",
    fgColor="17365D"
)

header_font = Font(
    color="FFFFFF",
    bold=True
)

alternate_fill = PatternFill(
    fill_type="solid",
    fgColor="F2F6FC"
)

priority_colors = {
    "High": "F4CCCC",
    "Medium": "FCE5CD",
    "Low": "D9EAD3",
    "Informational": "D9E2F3"
}


# 5. Export and format worksheets

with pd.ExcelWriter(
    PROFILING_REPORT_PATH,
    engine="openpyxl"
) as writer:

    for sheet_name, table in report_tables.items():

        table.to_excel(
            writer,
            sheet_name=sheet_name,
            index=False
        )

        worksheet = writer.sheets[sheet_name]

        worksheet.freeze_panes = "A2"
        worksheet.auto_filter.ref = worksheet.dimensions
        worksheet.sheet_view.showGridLines = False

        # Format header
        for cell in worksheet[1]:

            cell.fill = header_fill
            cell.font = header_font

            cell.alignment = Alignment(
                vertical="center",
                wrap_text=True
            )

        worksheet.row_dimensions[1].height = 32

        # Format rows
        for row in worksheet.iter_rows(min_row=2):

            for cell in row:

                cell.alignment = Alignment(
                    vertical="center"
                )

                if cell.row % 2 == 0:
                    cell.fill = alternate_fill

                if cell.data_type == "f":
                    # Treat source strings as text, not formulas
                    cell.data_type = "s"

                if isinstance(cell.value, float):
                    cell.number_format = "#,##0.0000"

        # Set practical column widths
        for column_cells in worksheet.columns:

            column_letter = get_column_letter(
                column_cells[0].column
            )

            maximum_length = max(
                len(str(cell.value))
                if cell.value is not None else 0
                for cell in column_cells[:201]
            )

            worksheet.column_dimensions[
                column_letter
            ].width = min(max(maximum_length + 3, 14), 65)

        # Highlight missing percentages
        if "missing_pct" in table.columns:

            col_idx = (
                table.columns.get_loc("missing_pct") + 1
            )

            letter = get_column_letter(col_idx)

            worksheet.conditional_formatting.add(
                f"{letter}2:{letter}{worksheet.max_row}",
                ColorScaleRule(
                    start_type="num",
                    start_value=0,
                    start_color="63BE7B",
                    mid_type="num",
                    mid_value=50,
                    mid_color="FFEB84",
                    end_type="num",
                    end_value=100,
                    end_color="F8696B"
                )
            )


    # 6. Format data quality priorities

    issues_sheet = writer.sheets["Quality Issues"]

    priority_col = (
        data_quality_issues_df.columns.get_loc("priority") + 1
    )

    treatment_col = (
        data_quality_issues_df.columns.get_loc(
            "proposed_treatment"
        ) + 1
    )

    for row_number in range(2, issues_sheet.max_row + 1):

        priority_cell = issues_sheet.cell(
            row=row_number,
            column=priority_col
        )

        if priority_cell.value in priority_colors:

            priority_cell.fill = PatternFill(
                fill_type="solid",
                fgColor=priority_colors[priority_cell.value]
            )

        issues_sheet.cell(
            row=row_number,
            column=treatment_col
        ).alignment = Alignment(
            wrap_text=True,
            vertical="center"
        )

        issues_sheet.row_dimensions[row_number].height = 35


    # 7. Add executive summary charts

    summary_sheet = writer.sheets["Executive Summary"]

    priority_counts = (
        data_quality_issues_df["priority"].value_counts()
    )

    summary_sheet["D2"] = "Quality Priority"
    summary_sheet["E2"] = "Observations"

    for row_number, priority in enumerate(
        ["High", "Medium", "Low", "Informational"],
        start=3
    ):

        summary_sheet.cell(
            row=row_number,
            column=4,
            value=priority
        )

        summary_sheet.cell(
            row=row_number,
            column=5,
            value=int(priority_counts.get(priority, 0))
        )

    for cell in summary_sheet[2][3:5]:
        cell.fill = header_fill
        cell.font = header_font

    summary_sheet.column_dimensions["D"].width = 22
    summary_sheet.column_dimensions["E"].width = 18

    quality_chart = BarChart()

    quality_chart.type = "bar"
    quality_chart.title = "Data Quality Observations"
    quality_chart.x_axis.title = "Number of observations"
    quality_chart.legend = None

    quality_chart.add_data(
        Reference(
            summary_sheet,
            min_col=5,
            min_row=2,
            max_row=6
        ),
        titles_from_data=True
    )

    quality_chart.set_categories(
        Reference(
            summary_sheet,
            min_col=4,
            min_row=3,
            max_row=6
        )
    )

    quality_chart.width = 17
    quality_chart.height = 9

    summary_sheet.add_chart(quality_chart, "D9")


# 8. Verify generated workbook

from openpyxl import load_workbook

workbook = load_workbook(
    PROFILING_REPORT_PATH,
    read_only=True
)

actual_sheets = workbook.sheetnames

assert set(actual_sheets) == set(report_tables)

print("COMMERCEIQ - DATA PROFILING REPORT")

print("\nWorkbook:", PROFILING_REPORT_PATH.name)

print("Worksheets created:", len(actual_sheets))

print("\nWORKSHEET INVENTORY")

display(pd.DataFrame([
    {
        "worksheet": name,
        "data_rows": len(table),
        "data_columns": len(table.columns)
    }
    for name, table in report_tables.items()
]))

workbook.close()

print(
    "\nFile size (KB):",
    round(PROFILING_REPORT_PATH.stat().st_size / 1024, 2)
)

print(
    "Report saved to:",
    PROFILING_REPORT_PATH.relative_to(PROJECT_ROOT)
)

print(
    "Report exists:",
    PROFILING_REPORT_PATH.is_file()
)

print("\nExcel profiling report generated successfully.")
```
